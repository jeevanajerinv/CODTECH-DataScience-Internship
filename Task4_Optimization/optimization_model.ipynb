```python
import pulp
import matplotlib.pyplot as plt

# Create optimization problem
model = pulp.LpProblem("Factory_Profit_Maximization", pulp.LpMaximize)

# Decision variables
product_A = pulp.LpVariable('Product_A', lowBound=0, cat='Integer')
product_B = pulp.LpVariable('Product_B', lowBound=0, cat='Integer')

# Objective function (maximize profit)
model += 50 * product_A + 40 * product_B, "Total Profit"

# Constraints
model += 2 * product_A + 1 * product_B <= 100, "Labor Constraint"
model += 1 * product_A + 1 * product_B <= 80, "Material Constraint"

# Solve model
model.solve()

# Results
print("Optimization Status:", pulp.LpStatus[model.status])
print("Optimal units of Product A:", product_A.varValue)
print("Optimal units of Product B:", product_B.varValue)
print("Maximum Profit: ₹", pulp.value(model.objective))

# Visualization
products = ['Product A', 'Product B']
values = [product_A.varValue, product_B.varValue]

plt.figure(figsize=(6,4))
plt.bar(products, values)
plt.title("Optimal Production Plan")
plt.xlabel("Products")
plt.ylabel("Units Produced")
plt.show()
```
